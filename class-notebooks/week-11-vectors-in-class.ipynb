```python
!pip install kagglehub
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mrisdal/fake-news")

print("Path to dataset files:", path)
```

```python
import pandas as pd

df = pd.read_csv('.cache/kagglehub/datasets/mrisdal/fake-news/versions/1/fake.csv')
```

```python
df.info()
```

```python
df['text']
```

```python
fake_news = df['text'].to_list()
```

```python
len(fake_news)
```

```python
fake_news[0]
```

```python
fake = []

for i in fake_news:
    try:
        j = i.split()
        for h in j:
            fake.append(h)
    except:
        pass
```

```python
len(fake)
```

```python
fake[:15]
```

```python
# removing punctuation
lower_no_punct = [word.lower() for word in fake if word.isalpha()]

# removing stop words
stops = stopwords.words('english')
no_stops = [word for word in lower_no_punct if word not in stops]

# lemmatizing
lemmatizer = WordNetLemmatizer()
clean_text = [lemmatizer.lemmatize(word, pos='n') for word in no_stops]
fake_cleaned = clean_text
```

```python
model = Word2Vec([fake_cleaned])
```

```python
type(model)
```

```python
model.wv.most_similar('money')
```

```python
model.wv.most_similar('trump')
```

```python
model.vector_size
```

```python
woolf_tokens = nltk.word_tokenize(r.text)
```

```python
type(woolf_tokens)
```

```python
type(woolf_tokens[0])
```

```python
woolf_text = woolf_tokens[868:-5]
```

```python
woolf_text[:10]
```

```python
# removing punctuation
lower_no_punct = [word.lower() for word in woolf_text if word.isalpha()]

# removing stop words
stops = stopwords.words('english')
no_stops = [word for word in lower_no_punct if word not in stops]

# lemmatizing
lemmatizer = WordNetLemmatizer()
clean_text = [lemmatizer.lemmatize(word, pos='n') for word in no_stops]
orlando = clean_text
```

```python
orlando[:10]
```

```python
o_model = Word2Vec([orlando], vector_size = 50)
```

```python
o_model.wv.most_similar(positive=["woman"], negative="man")
```
