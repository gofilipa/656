## cleaning
- what kind of predictions could we make from the data?
  - think about which of these columns can be used to guess about
    another column.
    - Example:
    - what states are likely to have "Worst" risk?
- the decisions we make about prediction will affect the way we clean
  and process the dataset.

```python
df["Bill Type"].value_counts()
```

```python
df["Bill Type Raw"] = df["Bill Type"]
```

```python
df.columns
```

```python
df["Bill Type"].str.split("/")
```

```python
df['Bill Type']
```

```python
df["bill_type_labels"] = df["Bill Type"].str.split("/").fillna("")
```

```python
df['bill_type_labels']
```

```python
def clean_labels(labels):
    return [label.strip() for label in labels]
```

```python
df["bill_type_labels"] = df["bill_type_labels"].apply(clean_labels)
```

```python
df['bill_type_labels']
```

## formatting data for training
- we will put the data into a matrix with positive labels (1) to mark the presence of a specific label
- do to so, we use the MultiLabelBinarizer class from the sklearn library. This class allows us to "binarize" the data, adding 1s where instances of the data are positive.

```python
from sklearn.preprocessing import MultiLabelBinarizer
```

```python
mlb = MultiLabelBinarizer()
```

```python
Y = mlb.fit_transform(df["bill_type_labels"])
```
