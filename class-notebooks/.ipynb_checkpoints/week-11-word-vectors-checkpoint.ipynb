## train a custom vector model

```python
# loading up the text, tokenizing the text
my_url = "http://gutenberg.net.au/ebooks02/0200331.txt"
r = requests.get(my_url)
source = r.text
woolf_tokens = nltk.word_tokenize(source)
o_text = woolf_tokens[872:-5]
o_text[:10]

# removing punctuation
lower_no_punct = [word.lower() for word in o_text if word.isalpha()]

# removing stop words
stops = stopwords.words('english')
no_stops = [word for word in lower_no_punct if word not in stops]

# lemmatizing
lemmatizer = WordNetLemmatizer()
clean_text = [lemmatizer.lemmatize(word, pos='n') for word in no_stops]
orlando = clean_text
```

```python
orlando[:10]
```

```python
# creating our language model using the Orlando text

model = Word2Vec([orlando], vector_size=500, window=20, min_count=1, workers=4)
```

# exploring *Orlando*'s vectors

# experimenting with `NLTK` methods

Run the below cells to see the "context window" for the search term. You might refer to the list of word vectors to brainstorm search terms.

You can change the word in quotes to one of your choice.

```python
# first, transform our text into format that we can use for text analysis with NLTK

orl = nltk.Text(orlando)
```

```python
# use the "concordance" method for getting context of a search word

orl.concordance("woman")
```

```python
# try it with another word

orl.concordance("soft")
```

```python
orl.collocations()
```

```python
orl.dispersion_plot(['woman', 'man', 'tree', 'sailor', 'goose', 'princess'])
```
